```python
df_clean = df.dropna()

df_clean.isna().sum()
```

```python
fig = px.histogram(
    df_clean,
    x='price',
    nbins=50,
    title='Distribución de precios'
)

fig.show()
```

```python
fig = px.histogram(
    df_clean,
    x='odometer',
    nbins=50,
    title='Distribución del kilometraje'
)

fig.show()
```

```python
df_clean['type'].value_counts()
```

```python
fig = px.bar(
    df_clean['type'].value_counts(),
    title='Cantidad de vehículos por tipo'
)

fig.show()
```

```python
df_clean['condition'].value_counts()
```

```python
fig = px.bar(
    df_clean['condition'].value_counts(),
    title='Condición de los vehículos'
)

fig.show()
```

```python
fig = px.scatter(
    df_clean,
    x='odometer',
    y='price',
    title='Precio vs Kilometraje'
)

fig.show()
```

```python
fig = px.scatter(
    df_clean,
    x='model_year',
    y='price',
    title='Precio según año del vehículo'
)

fig.show()
```

```python
fig = px.box(
    df_clean,
    x='type',
    y='price',
    title='Precio por tipo de vehículo'
)

fig.show()
```
